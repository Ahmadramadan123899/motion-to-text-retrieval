```python
import json
history = json.load(open('checkpoints/val_run/history.json'))
best = max(history, key=lambda h: h['score'])
print('best epoch:', best['epoch'], ' score:', round(best['score'], 4))
```

## 2. Final model on all the data

```python
!python train.py --full --epochs {best['epoch']} --out-dir checkpoints/final
```
